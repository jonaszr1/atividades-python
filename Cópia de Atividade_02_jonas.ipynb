```python
#Questão 01
numero = int(input("Digite um número:"))
if numero > 0:
  print("Positivo")
elif numero <0:
  print("Negativo")
else:
  print("Zero")
```

```python
#Questão 02
a = int(input("Digite o primeiro número:"))
b = int(input("Digite o segundo número:"))

if a > b :
  print("Maior:", a)
elif b > a :
  print("Maior:", b)
else:
  print("Os números são iguais")
```

```python
#Questão 03
soma = 0
for i in range (5) :
  numero = int(input("Digite um número:"))
  soma = soma + numero
print("Soma:", soma)
```

```python
#Questão 04
n = int(input("Digite um número:"))

if n % 2 == 0 :
  print("O número é par")
elif n % 2 != 0 :
  print("O número é impar")
```

```python
#Questão 05
nota = int(input("Digite a nota:"))

if nota >= 7 :
  print("Aprovado")
elif nota >= 5 and nota < 7 :
  print("Recuperação")
else :
  print("Reprovado")
```

```python
#Questão 06
a = int(input("Digite o primeiro número:"))
b = int(input("Digite o segundo número:"))
c = int(input("Digite o terceiro número:"))

if a > b and a > c :
  print("Maior:", a)
elif b > a and b > c :
  print("Maior:", b)
elif c > a and c > b :
  print("Maior:", c)
```
